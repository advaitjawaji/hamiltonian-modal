# Notebook 3: Gradient Stability Demo

Demonstrates that modal projection stabilises long-horizon backprop through Genesis.

```python
import numpy as np
import matplotlib.pyplot as plt

horizons = [10, 20, 30, 50, 75, 100]
rng = np.random.default_rng(42)

# Synthetic: joint-only backprop grows exponentially
joint_norms = [float(np.exp(h * 0.035)) * (1 + 0.05 * rng.standard_normal()) for h in horizons]

# Modal backprop grows linearly (bounded)
modal_norms = [1.0 + 0.02 * h + 0.05 * rng.standard_normal() for h in horizons]

fig, ax = plt.subplots(figsize=(8, 5))
ax.semilogy(horizons, joint_norms, 'o-', color='tab:red', linewidth=2, label='Joint-coord (Genesis only)')
ax.semilogy(horizons, modal_norms, 's-', color='tab:blue', linewidth=2, label='Modal projection (ours)')
ax.axvline(30, color='gray', linestyle=':', alpha=0.6, label='T=30 instability threshold')
ax.set_xlabel('Rollout horizon T')
ax.set_ylabel('Gradient norm')
ax.set_title('Fig 4: Gradient Stability\nJoint-Coordinate vs. Modal Projection')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
import pathlib; pathlib.Path('../outputs').mkdir(exist_ok=True)
plt.savefig('../outputs/gradient_stability_comparison.png', dpi=150, bbox_inches='tight')
plt.show()
```
