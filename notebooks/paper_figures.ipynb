# Paper Figures Notebook

Produces all 8 paper figures. This is the interactive version of `scripts/reproduce_paper_figures.py`.

```python
import subprocess, sys
result = subprocess.run(
    [sys.executable, '../scripts/reproduce_paper_figures.py', '--output-dir', '../outputs/paper_figures'],
    capture_output=True, text=True
)
print(result.stdout)
if result.returncode != 0:
    print('STDERR:', result.stderr)
```

```python
import pathlib
from IPython.display import Image, display
for fig in sorted(pathlib.Path('../outputs/paper_figures').glob('fig*.png')):
    print(f'--- {fig.name} ---')
    display(Image(str(fig), width=700))
```
