# Notebook 1: G1 Modal Decomposition Visualization

Visualizes the top 20 G1 mode shapes and their natural frequencies.

```python
import numpy as np
import matplotlib.pyplot as plt
from hamiltonian_modal.modal.decomposition import compute_modal_decomposition, is_m_orthonormal
from hamiltonian_modal.modal.stiffness import assemble_stiffness_matrix

# Synthetic mass matrix (replace with Pinocchio CRBA for real G1)
rng = np.random.default_rng(42)
n = 23
A = rng.standard_normal((n, n))
M = A @ A.T + 5 * np.eye(n)

# Stiffness from PD gains + gearbox + bending
K = assemble_stiffness_matrix(model=None, q_ref=np.zeros(n))

eigenvalues, mode_shapes = compute_modal_decomposition(M, K, n_modes=20)
omega_hz = np.sqrt(eigenvalues) / (2 * np.pi)

print(f'Mass orthonormal: {is_m_orthonormal(mode_shapes, M)}')
print(f'Natural frequencies (Hz): {omega_hz[:5].round(2)}')
```

```python
fig, axes = plt.subplots(4, 5, figsize=(16, 10))
for i, ax in enumerate(axes.flat):
    ax.bar(range(n), mode_shapes[:, i], color='tab:blue', alpha=0.7)
    ax.set_title(f'Mode {i+1}\n{omega_hz[i]:.1f} Hz', fontsize=9)
    ax.set_xticks([])
    ax.set_yticks([])
fig.suptitle('G1 Mode Shapes (top 20 lowest-frequency modes)', fontsize=13)
plt.tight_layout()
plt.savefig('../outputs/g1_modes_visualization.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(9, 4))
ax.stem(range(1, 21), omega_hz, markerfmt='o', linefmt='tab:blue', basefmt='k-')
ax.set_xlabel('Mode index')
ax.set_ylabel('Natural frequency (Hz)')
ax.set_title('G1 Natural Frequency Spectrum')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
