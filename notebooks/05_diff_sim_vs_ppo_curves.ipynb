# Notebook 5: Diff-Sim vs. PPO Learning Curves

Produces Fig 5 (Headline A: ≥10× sample efficiency).

```python
import numpy as np
import matplotlib.pyplot as plt
from benchmarks.diffsim_effbench.runner import run_benchmark

reports = run_benchmark(n_steps=1000, seed=42)
for r in reports:
    print(f'{r.method_name}: final={r.final_reward:.3f}, 80%-threshold step={r.steps_to_80pct_threshold}')
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
colors = ['tab:blue', 'tab:red']
for r, color in zip(reports, colors):
    ax.plot(np.arange(len(r.reward_curve)), r.reward_curve, color=color, linewidth=2, label=r.method_name)
ax.axhline(0.8, color='gray', linestyle='--', alpha=0.5, label='80% quality threshold')
ax.set_xlabel('Training steps')
ax.set_ylabel('Task reward (normalised)')
ax.set_title('Fig 5: Sample Efficiency — Diff-Sim vs. PPO (Headline A)')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
import pathlib; pathlib.Path('../outputs').mkdir(exist_ok=True)
plt.savefig('../outputs/diff_sim_vs_ppo.png', dpi=150, bbox_inches='tight')
plt.show()
```
